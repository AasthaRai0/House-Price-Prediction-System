```python
#DATA PREPROCESSING

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(user_agent="my_geocoder")
reverse = RateLimiter(geolocator.reverse, max_retries=3, error_wait_seconds=1.0)
location = reverse("37.88, -122.23")
```

```python
geolocator.reverse("37.88, -122.23")[0]
```

```python
def location(cord):
  latitude = str(cord[0])
  longitude = str(cord[1])

  res = geolocator.reverse(latitude + ',' + longitude)

  if res is not None:
    loc_data = res.raw.get('address', {})
  else:
    loc_data = {} 
  road = loc_data.get('road', None)
  county = loc_data.get('county', None)
  city = loc_data.get('city', None)
  loc_update['Road'].append(road)
  loc_update['County'].append(county)
  loc_update['City'].append(city)
```

```python
import pickle
```

```python
import pickle
import time as t  # import time under a completely safe alias
from geopy.exc import GeocoderTimedOut, GeocoderUnavailable

# Load existing progress if available, otherwise start fresh
try:
  loc_update = pickle.load(open('loc_update.pkl', 'rb'))
except FileNotFoundError:
  loc_update = {'County': [], 'Road': [], 'City': []}

# Find out how many rows are already saved
start_index = len(loc_update['County'])
print(f'Resuming from index {start_index}...')

for i, cord in enumerate(df.iloc[:, 6::-1].values):
  if i < start_index:
    continue  # Skip already processed rows

  success = False
  while not success:
    try:
      location(cord)
      success = True
    except (GeocoderTimedOut, GeocoderUnavailable, TimeoutError):
      print(f'Timeout at index {i}. Retrying in 5 seconds...')
      t.sleep(5)  # Use t.sleep instead of time.sleep

  # Save progress incrementally every 100 rows
  if i % 100 == 0:
    pickle.dump(loc_update, open('loc_update.pkl', 'wb'))
    print(f'Saved progress up to index {i}')
    t.sleep(1)

# Final save when complete
pickle.dump(loc_update, open('loc_update.pkl', 'wb'))
print('Geocoding complete and saved!')
```
